```python
import pandas as pd

data = {
    "Day": ["Monday","Monday","Tuesday","Tuesday","Wednesday","Wednesday","Thursday","Thursday","Friday","Friday"],
    "Coffee Type": ["Latte","Espresso","Latte","Espresso","Latte","Espresso","Latte","Espresso","Latte","Espresso"],
    "Units Sold": [15,25,20,30,25,35,30,40,35,45],
    "Price": [350,300,350,300,350,300,350,300,350,300],
    "Payment": ["Cash","Card","Cash","Card","Cash","Card","Cash","Card","Cash","Card"]
}

coffee = pd.DataFrame(data)

coffee["Total Sales"] = coffee["Units Sold"] * coffee["Price"]

coffee.to_csv("coffee_sales.csv", index=False)

print(coffee)
```

```python
coffee.sort_values(by=["Units Sold", "Coffee Type"], ascending=False)
```

```python
coffee.sort_values(by=["Units Sold", "Coffee Type"], ascending=(0,1))
```
