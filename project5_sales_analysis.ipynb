```python
#PROJECT 5
import pandas as pd

data = {
    "Date": ["2026-07-01","2026-07-02","2026-07-03","2026-07-04","2026-07-05","2026-07-06","2026-07-07","2026-07-08","2026-07-09","2026-07-10"],
    "Product":["laptop","mouse","keyboard","monitor","printer","table","laptop","keyboard","printer","monitor"],
     "Category":["Electronics","Accessories","Accessories","Electronics","Electronics","Electronics","Accessories","Electronics","Accessories","Electronics"],
    "Price":[80000,1500,3500,30000,25000,45000,80000,1500,30000,3500],
    "Quantity":[2,5,3,2,1,2,1,4,3,2],
    "City":["Lahore","Karachi","Islamabad","Rawalpindi","Multan","Lahore","Karachi","Islamabad","Rawalpindi","Multan"],
    "Salesperson":["Ali","Ahmed","Hassan","Ali","Ahmed","Hassan","Ali","Ahmed","Hassan","Ali"]

    }
sales = pd.DataFrame(data)

sales.to_csv("sales.csv", index=False)

print(sales.head())

print(sales.tail())

print(sales.info())

print(sales.describe())

print(sales.isnull().sum())

print(sales.duplicated().sum())

print(sales["Product"].value_counts())

print(sales.groupby("Category")["Price"].mean())

print(sales.groupby("City")["Quantity"].sum())

print(sales.sort_values("Price",ascending=False))



print(sales.sort_values("Quantity",ascending=False))

sales["Total Sales"] = sales["Price"] * sales["Quantity"]
print(sales)

print(sales.groupby("Salesperson")["Total Sales"].sum())

print(sales.groupby("Category")["Total Sales"].sum())

print(sales.groupby("City")["Total Sales"].sum())

print(sales.groupby("Product")["Total Sales"].sum())

print(sales)

import matplotlib.pyplot as plt
product_sales = sales.groupby("Product")["Total Sales"].sum()
plt.bar(product_sales.index,product_sales.values)
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.title("Total Sales by Product")
plt.show()

category_sales = sales.groupby("Category")["Total Sales"].sum()
plt.pie(category_sales.values,labels=category_sales.index,autopct="%1.1f%%")
plt.title("Total Sales by Category")
plt.show()

city_sales = sales.groupby("City")["Total Sales"].sum()
plt.scatter(city_sales.index,city_sales.values)
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.title("Total Sales by City")
plt.show()

category_sales = sales.groupby("Category")["Total Sales"].sum()
plt.bar(category_sales.index,category_sales.values)
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.title("Total Sales by Category")
plt.show()

salesperson_sales = sales.groupby("Salesperson")["Total Sales"].sum()
plt.pie(salesperson_sales.values,labels=salesperson_sales.index,autopct="%1.1f%%")
plt.title("Total Sales by Salesperson")
plt.show()

category_sales.plot(kind="bar")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.title("Total Sales by Category")
plt.show()

category_sales.plot(kind="pie",autopct="%1.1f%%")
plt.title("Total Sales by Category")
plt.show()


```
