```python
import pandas as pd


data={
"Employee id":["EMP001","EMP002","EMP003","EMP004","EMP005","EMP006","EMP007","EMP008","EMP009","EMP010"],
"Name":["Ali","Ayesha","Hamza","Zainab","Zain","komal","Alina", "Tajmbal","Asia","Izza"],
"Department" :[ "IT","HR","Finance","Sales", "Marketing","IT","HR","Finance","sales", "Marketing"],
"City":["Rawalpindi", "Islamabad", "chakwal"," Multan", "Lahore", "Karachi", "Islamabad", "Lahore","Karachi","Multan"],
"Age" :[20,22,21,23,22,20,22,23,22,21],
"Salary":[2000,3000,22000,21000,21000,20000,21000,22000,20000,22000],
"Experience(year)":[2,5,1,7,3,3,4,5,6,4],
"Gender":["Male", "Female","Male", "Female","Male","Female", "Female","Male", "Female","Female"]
}
Employee = pd.DataFrame(data)

Employee["Department"] = Employee["Department"].str.title()

Employee.to_csv("Employee_data.csv", index=False)

print(Employee)

print (Employee.head())

print(Employee.tail())

print (Employee.shape)

Employee.info()

print (Employee.describe())

print (Employee.value_counts())

print (Employee.groupby("Department")["Salary"].mean())

print (Employee.query("Salary > 20000"))

print(Employee.groupby("City")["Salary"].sum())

print(Employee.groupby("Gender").size())

print(Employee["Salary"].max())

print(Employee["Salary"].min())

print(Employee.sort_values("Salary", ascending=False))

Employee.to_csv("Employee_Data_Analysis.csv", index=False)

#Bonus
Employee["Bonus"] = Employee["Salary"] * 0.10
print(Employee)

#save CSV
Employee.to_csv("Employee_Data_Analysis.csv", index=False)
```
